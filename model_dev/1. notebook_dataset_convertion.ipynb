```python
def normalize_image(images):
    image_list = []
    
    for image in images:
        img = cv2.imread(image)
        img = cv2.resize(img, (224, 224))
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        
        image_list.append(img.astype("float32"))
    
    return np.array(image_list)
```

```python
X0_train = normalize_image(Acral_Lentiginous_Melanoma_images_train)
X1_train = normalize_image(blue_finger_images_train)
X2_train = normalize_image(clubbing_images_train)
X3_train = normalize_image(Healthy_Nail_images_train)
X4_train = normalize_image(Onychogryphosis_images_train)
X5_train = normalize_image(pitting_images_train)
```

```python
X0_test = normalize_image(Acral_Lentiginous_Melanoma_images_test)
X1_test = normalize_image(blue_finger_images_test)
X2_test = normalize_image(clubbing_images_test)
X3_test = normalize_image(Healthy_Nail_images_test)
X4_test = normalize_image(Onychogryphosis_images_test)
X5_test = normalize_image(pitting_images_test)
```

```python
X0_train_ds2 = normalize_image(Healthy_Nail_images_train_ds2)
X1_train_ds2 = normalize_image(Onychomycosis_images_train_ds2)
X2_train_ds2 = normalize_image(psoriasis_images_train_ds2)
```

```python
X0_test_ds2 = normalize_image(Healthy_Nail_images_test_ds2)
X1_test_ds2 = normalize_image(Onychomycosis_images_test_ds2)
X2_test_ds2 = normalize_image(psoriasis_images_test_ds2)
```

```python
healthy_nails_train_combined = np.concatenate([X3_train,X0_train_ds2])
healthy_nails_test_combined = np.concatenate([X3_test,X0_test_ds2])
```

```python
y0_train = np.full(len(healthy_nails_train_combined),0).astype(np.int32)
y1_train = np.full(len(X0_train),1).astype(np.int32)
y2_train = np.full(len(X1_train),2).astype(np.int32)
y3_train = np.full(len(X2_train),3).astype(np.int32)
y4_train = np.full(len(X4_train),4).astype(np.int32)
y5_train = np.full(len(X5_train),5).astype(np.int32)
y6_train = np.full(len(X1_train_ds2),6).astype(np.int32)
y7_train = np.full(len(X2_train_ds2),7).astype(np.int32)
```

```python
y0_test = np.full(len(healthy_nails_test_combined),0).astype(np.int32)
y1_test = np.full(len(X0_test),1).astype(np.int32)
y2_test = np.full(len(X1_test),2).astype(np.int32)
y3_test = np.full(len(X2_test),3).astype(np.int32)
y4_test = np.full(len(X4_test),4).astype(np.int32)
y5_test = np.full(len(X5_test),5).astype(np.int32)
y6_test = np.full(len(X1_test_ds2),6).astype(np.int32)
y7_test = np.full(len(X2_test_ds2),7).astype(np.int32)
```
